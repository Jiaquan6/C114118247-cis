```python
import csv
import json
import math
import random

# 設定隨機種子以確保結果可複現
random.seed(42)

data = []
n = 60  # 生成 60 筆資料

for i in range(1, n + 1):
    student_id = i
    study_hours = round(random.uniform(1.0, 10.0), 1)
    sleep_hours = round(random.uniform(5.0, 9.0), 1)
    screen_time_hours = round(random.uniform(2.0, 8.0), 1)
    attendance_rate = random.randint(60, 100)

    # 依圖片中的邏輯計算分項
    study_part = 3.0 * study_hours
    sleep_part = 1.5 * sleep_hours
    screen_part = -1.0 * screen_time_hours
    attendance_part = 0.15 * attendance_rate
    random_error = random.gauss(0, 5)  # 常態分佈常數誤差 (mean=0, std=5)

    # 計算原始分數與最終 quiz_score
    raw_score = (
        50
        + study_part
        + sleep_part
        + screen_part
        + attendance_part
        + random_error
    )

    # 限制分數在 0 到 100 之間並取整數
    quiz_score = int(round(min(max(raw_score, 0), 100)))

    data.append(
        {
            "student_id": student_id,
            "study_hours": study_hours,
            "sleep_hours": sleep_hours,
            "screen_time_hours": screen_time_hours,
            "attendance_rate": attendance_rate,
            "quiz_score": quiz_score,
        }
    )

# 顯示前 5 筆資料確認
print(f"成功生成 {len(data)} 筆學生模擬資料！\n")
print(
    f"{'student_id':<12} {'study_hours':<13} {'sleep_hours':<12} {'screen_time_hours':<18} {'attendance_rate':<17} {'quiz_score':<10}"
)
print("-" * 85)
for row in data[:60]:  # 顯示 60 筆
    print(
        f"{row['student_id']:<12} {row['study_hours']:<13} {row['sleep_hours']:<12} {row['screen_time_hours']:<18} {row['attendance_rate']:<17} {row['quiz_score']:<10}"
    )

# 匯出資料為 CSV 備份
with open("mission8_data.csv", "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.DictWriter(f, fieldnames=list(data[0].keys()))
    writer.writeheader()
    writer.writerows(data)

print("\n資料已成功儲存為 mission8_data.csv！")
```
